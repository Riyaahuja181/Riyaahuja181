```python

import random

def guess_the_number_multiplayer():
    number_to_guess = random.randint(1, 100)
    max_attempts = 10
    players = []
    player_attempts = {}
    hints_used = {}

    num_players = int(input("How many players? (1-7): "))
    for i in range(num_players):
        name = input(f"Enter name for Player {i+1}: ")
        players.append(name)
        player_attempts[name] = 0
        hints_used[name] = False

    print("\nI'm thinking of a number between 1 and 100.")
    print(f"Each player gets up to {max_attempts} attempts. You can use one hint per player.")

    winner = None
    while True:
        for player in players:
            if player_attempts[player] >= max_attempts:
                print(f"{player} has used all their attempts.")
                continue

            print(f"\n{player}'s turn. Attempts left: {max_attempts - player_attempts[player]}")

            action = input("Type 'guess' to guess or 'hint' to get a hint: ").lower()

            if action == 'hint':
                if not hints_used[player]:
                    hints_used[player] = True
                    if number_to_guess % 2 == 0:
                        print("Hint: The number is even. Or is it??")
                    else:
                        print("Hint: The number is odd. Or is it?")
                else:
                    print("You already used your hint.")
                # Player still gets to guess after hint
                action = 'guess'

            if action == 'guess':
                try:
                    guess = int(input("Enter your guess: "))
                    player_attempts[player] += 1

                    if guess < number_to_guess:
                        print("Too low. Idiot!")
                    elif guess > number_to_guess:
                        print("Too high. Ah well?")
                    else:
                        winner = player
                        break

                except ValueError:
                    print("Invalid input. Please enter an integer.")

        if winner:
            print(f"\nCongratulations {winner}! You guessed the number correctly!")
            break

        # Check if all players exhausted attempts
        if all(player_attempts[p] >= max_attempts for p in players):
            print("\nGame over! No one guessed the number.")
            print(f"The number was: {number_to_guess}")
            break

    # Show attempts leaderboard
    print("\n--- Attempts leaderboard ---")
    for player, attempts in player_attempts.items():
        print(f"{player}: {attempts} attempts")

if __name__ == "__main__":
    guess_the_number_multiplayer()
```
